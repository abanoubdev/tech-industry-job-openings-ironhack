## Label Encoding for categorical columns


```python

df = pd.read_csv('../DataSet/cleaned_job_data.csv')

categorical_cols = [
    'Job_Title', 'Industry', 'Job_Status', 'Required_Education', 
    'Location', 'Risk_Category', 'Seniority', 'AI_Impact_Level'
]

encoders = {}

for col in categorical_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    encoders[col] = le

joblib.dump(encoders, '../DataSet/Encoder/label_encoders.pkl')
df.head()
```

## Use StandardScaler to fit large numbers like Median_Salary_(USD) for ML model for higher predictions

```python
import joblib

numerical_cols = [
    'Median_Salary_(USD)', 'Experience_Required_(Years)', 'Job_Openings_(2024)',
    'Projected_Openings_(2030)', 'Remote_Work_Ratio', 'Automation_Risk',
    'Gender_Diversity', 'Growth_Rate', 'AI_Exposure_Index', 
    'Tech_Growth_Factor', 'Automation_Probability_2030'
]

scaler = StandardScaler()

df[numerical_cols] = scaler.fit_transform(df[numerical_cols])

joblib.dump(scaler, '../DataSet/Scaler/scaler.pkl')
df
```

```python
is_high_encoded_val = encoders['AI_Impact_Level'].transform(['High'])[0]

df['Is_High_Risk'] = (df['AI_Impact_Level'] == is_high_encoded_val).astype(int)

df = df.drop(columns=['AI_Impact_Level'])

print("Feature Engineering complete. Target column is now 'Is_High_Risk'.")
print("High Risk Jobs:", df['Is_High_Risk'].sum())
print("Total Jobs:", len(df))
```
